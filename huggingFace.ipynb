```python
# 406

import torch
from transformers import AutoModelForSequenceClassification,AutoTokenizer,pipeline,DistilBertForSequenceClassification,DistilBertTokenizer

model_name1="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
mytokenizer1=DistilBertTokenizer.from_pretrained(model_name1)
mymodel1=DistilBertForSequenceClassification.from_pretrained(model_name1)

classifier = pipeline(
    "sentiment-analysis",
    model=mymodel1, 
    tokenizer=mytokenizer1
)

result = classifier("I was not happy with the boking service.")
print(result)

```

```python
model_name2 = "nlptown/bert-base-multilingual-uncased-sentiment"

mymodel2=AutoModelForSequenceClassification.from_pretrained(model_name2)
mytokenizer2=AutoTokenizer.from_pretrained(model_name2)

classifier = pipeline(
    "sentiment-analysis",
    model=mymodel2,
    tokenizer=mytokenizer2
)

result = classifier("The product was okay, not amazing but works.")
print(result)
```

```python
print(result)
```

```python
# A multilingual test batch
multilingual_reviews = [
    "¡Este producto es absolutamente increíble, me encanta!", # Spanish: 5 Stars
    "C'est une catastrophe total, rien ne fonctionne.",        # French: 1 Star
    "Das Essen war ganz okay, aber der Service war langsam.",  # German: 3 Stars
    "Very fast delivery, highly recommend to everyone."        # English: 5 Stars
]

# Run the batch through your active pipeline
results = classifier(multilingual_reviews)

# View the cross-language grades
print("--- MULTILINGUAL RESULTS ---")
for text, res in zip(multilingual_reviews, results):
    print(f"\nText:  \"{text}\"")
    print(f"Grade: {res['label']} (Confidence: {res['score']:.4f})")
```
